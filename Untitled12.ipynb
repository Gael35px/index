# 🌐 Traductor Español a Inglés + Análisis de Sentimientos

## Objetivo
Traducir palabras del español al inglés usando un diccionario en Python y analizar el sentimiento de una frase.

## Descripción
Este notebook crea un traductor básico y realiza análisis de sentimientos con TextBlob.

## Diccionario de traducción

Se define un diccionario en Python con palabras en español y su traducción al inglés.

```python
dic = {
    "hola": "hello",
    "mundo": "world",
    "adios": "goodbye",
    "gracias": "thank you",
    "por favor": "please",
    "amor": "love",
    "feliz": "happy",
    "triste": "sad",
    "perro": "dog",
    "gato": "cat",
    "casa": "house",
    "escuela": "school"
}
```

## Función de traducción

Se utiliza el método <b>.get()</b> para evitar errores si la palabra no existe.

```python
def translate(word):
    return dic.get(word.lower(), "Palabra no encontrada")
```

## Prueba de traducción

Se prueba la función con una palabra.

```python
print("Traducción de 'hola':", translate("hola"))
```

## Traducción de múltiples palabras

Se muestran más de 10 palabras traducidas.

```python
for palabra in dic:
    print(palabra, "->", translate(palabra))
```

## Traducción interactiva

El usuario puede ingresar una palabra para traducirla.

```python
palabra_usuario = input("Ingresa una palabra en español: ")
print("Traducción:", translate(palabra_usuario))
```
