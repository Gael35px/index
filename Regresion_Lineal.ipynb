```python
# Crear variables dummy (one-hot encoding) para la columna categórica
dummies = pd.get_dummies(datos_na["ocean_proximity"], dtype=int)
```

```python
# Unir las variables dummy al DataFrame principal
datos_na = datos_na.join(dummies)
```

```python
# Eliminar la columna original 'ocean_proximity' para evitar redundancia
datos_na = datos_na.drop(["ocean_proximity"], axis=1)
```

```python
# Separar variables independientes (X) y variable dependiente (y)
X = datos_na.drop(["median_house_value"], axis=1)
y = datos_na["median_house_value"]
```

```python
# Dividir los datos en conjunto de entrenamiento (80%) y prueba (20%)
from sklearn.model_selection import train_test_split

X_ent, X_pru, y_ent, y_pru = train_test_split(X, y, test_size=.2)
```

```python
# Importar y crear el modelo de regresión lineal
from sklearn.linear_model import LinearRegression

modelo=LinearRegression()
```

```python
# Entrenar el modelo con los datos de entrenamiento
modelo.fit(X_ent, y_ent)
```

```python
# Generar predicciones usando los datos de prueba
predicciones = modelo.predict(X_pru)
```

```python
# Crear una comparativa entre valores predichos y valores reales (aunque solo se muestran predicciones)
comparativa = {"Prediccion": predicciones, "Valor Real": y_pru}
pd.DataFrame(predicciones)
```

```python
# Evaluar el modelo usando R^2 en entrenamiento y prueba
print(modelo.score (X_ent, y_ent))
print(modelo.score (X_pru, y_pru))
```

```python
# Calcular el error cuadrático medio (MSE)
from sklearn.metrics import mean_squared_error
import numpy as np
mse = mean_squared_error(y_pru, predicciones)
```

```python
# Mostrar el valor del MSE
mse
```

```python
# Mostrar el valor del RMSE
rmse = np.sqrt(mse)
```
