```python
# ================================
# 1. IMPORTAR LIBRERÍAS
# ================================
from googleapiclient.discovery import build
from transformers import pipeline
from collections import Counter
import matplotlib.pyplot as plt

# ================================
# 2. FUNCIÓN PARA EXTRAER COMENTARIOS
# ================================
def get_youtube_comments(video_id, api_key, max_results=100):
    youtube = build('youtube', 'v3', developerKey=api_key)

    comments = []  # Lista para guardar comentarios
    next_page_token = None
    total = 0

    # Obtener comentarios hasta llegar al límite
    while total < max_results:
        request = youtube.commentThreads().list(
            part='snippet',
            videoId=video_id,
            maxResults=100,
            pageToken=next_page_token,
            textFormat='plainText'
        )

        response = request.execute()

        # Extraer texto de comentarios
        for item in response['items']:
            comment = item['snippet']['topLevelComment']['snippet']['textDisplay']
            comments.append(comment)
            total += 1

        # Revisar si hay más páginas
        if 'nextPageToken' in response:
            next_page_token = response['nextPageToken']
        else:
            break

    return comments


# ================================
# 3. CONFIGURACIÓN
# ================================
API_KEY = "AIzaSyBZm_mkEq-eSk3ZqScAde7IujbpV0BrQQ8"  # 🔴
video_id = "esMjQ75SRuI"  # ID del video que elegiste
max_comentarios = 200


# ================================
# 4. EXTRAER COMENTARIOS
# ================================
comentarios = get_youtube_comments(video_id, API_KEY, max_results=max_comentarios)

print(f"Se obtuvieron {len(comentarios)} comentarios")


# ================================
# 5. LIMPIEZA DE DATOS
# ================================
# Eliminar comentarios vacíos
comentarios = [c for c in comentarios if c.strip() != ""]

# Limitar longitud (mejora rendimiento del modelo)
comentarios = [c[:512] for c in comentarios]


# ================================
# 6. ANÁLISIS DE SENTIMIENTOS
# ================================
# Cargar modelo de 1 a 5 estrellas
sentiment_pipeline = pipeline(
    "sentiment-analysis",
    model="nlptown/bert-base-multilingual-uncased-sentiment"
)

# Analizar comentarios
resultados = sentiment_pipeline(comentarios)

# Convertir etiquetas a números (1 a 5)
estrellas = [int(r['label'][0]) for r in resultados]


# ================================
# 7. CONTAR RESULTADOS
# ================================
conteo = Counter(estrellas)

print("Conteo de sentimientos:")
print(conteo)


# Mostrar comentarios del 5 al 10
print("\nEjemplo de comentarios (del 5 al 10):")
for i, c in enumerate(comentarios[5:10], start=6):
    print(f"{i}. {c}\n")

# Mostrar comentarios del 5 al 10 con sus estrellas
print("\nEjemplo de comentarios (del 5 al 10) con sentimiento:")

for i, (c, e) in enumerate(zip(comentarios[5:10], estrellas[5:10]), start=6):
    print(f"{i}. ({e}⭐) {c}\n")
# ================================
# 8. GRÁFICA FINAL
# ================================
# Asegurar las 5 categorías
valores = [conteo.get(i, 0) for i in range(1, 6)]

plt.bar(['1⭐','2⭐','3⭐','4⭐','5⭐'], valores)

plt.title("Análisis de Sentimientos de Comentarios")
plt.xlabel("Clasificación (Estrellas)")
plt.ylabel("Cantidad de Comentarios")

plt.show()


# ================================
# 9. PORCENTAJE DE SENTIMIENTOS
# ================================
total = len(estrellas)

print("\nPorcentaje de sentimientos:")

for i in range(1, 6):
    porcentaje = (conteo.get(i, 0) / total) * 100
    print(f"{i}⭐: {porcentaje:.2f}%")
```
