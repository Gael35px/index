```python

# 2. Importaciones
import pandas as pd
from docx import Document
from docx.shared import Inches, Pt
from datetime import datetime
import os
import matplotlib.pyplot as plt

# Configuración
EXCEL_FILE = "datos_alumnos.xlsx"
TEMPLATE_FILE = "Plantilla.docx"
OUTPUT_DIR = "cartas_generadas"
os.makedirs(OUTPUT_DIR, exist_ok=True)

# Cargar datos
df = pd.read_excel(EXCEL_FILE)
print("Datos cargados:")
print(df.head())

# Fecha actual
fecha_actual = datetime.now().strftime("%d/%m/%Y")

# Funciones auxiliares
def set_document_font(doc, font_size=10, font_name='Arial'):
    """Ajusta el tamaño de fuente de todo el documento"""
    style = doc.styles['Normal']
    style.font.size = Pt(font_size)
    style.font.name = font_name
    # Aplicar estilo a todos los párrafos del cuerpo
    for paragraph in doc.paragraphs:
        paragraph.style = style
    # Aplicar a las celdas de tablas
    for table in doc.tables:
        for row in table.rows:
            for cell in row.cells:
                for paragraph in cell.paragraphs:
                    paragraph.style = style

def replace_text_in_paragraph(paragraph, old_text, new_text):
    if old_text in paragraph.text:
        for run in paragraph.runs:
            if old_text in run.text:
                run.text = run.text.replace(old_text, new_text)

def replace_text_in_doc(doc, replacements):
    for paragraph in doc.paragraphs:
        for old, new in replacements.items():
            replace_text_in_paragraph(paragraph, old, new)

def fill_table_row(doc, row_data):
    table = doc.tables[0]
    if len(table.rows) > 1:
        data_row = table.rows[1]
        cells = data_row.cells
        cells[0].text = str(row_data["Alumno"])
        cells[1].text = str(row_data["Español"])
        cells[2].text = str(row_data["Matemáticas"])
        cells[3].text = str(row_data["Ciencias"])
        cells[4].text = str(row_data["Historia"])
        cells[5].text = str(row_data["Biología"])
        cells[6].text = str(row_data["Promedio"])
    else:
        row_cells = table.add_row().cells
        row_cells[0].text = str(row_data["Alumno"])
        row_cells[1].text = str(row_data["Español"])
        row_cells[2].text = str(row_data["Matemáticas"])
        row_cells[3].text = str(row_data["Ciencias"])
        row_cells[4].text = str(row_data["Historia"])
        row_cells[5].text = str(row_data["Biología"])
        row_cells[6].text = str(row_data["Promedio"])

def get_recommendation_paragraph(promedio, nombre):
    if promedio >= 9:
        return f"El promedio de {nombre} es mayor de 9, es un alumno ejemplar. ¡Felicidades!"
    elif 6 <= promedio < 9:
        return f"El promedio de {nombre} es entre 6 y 8, su hijo(a) lo hace bien, cumple con todo pero siempre se puede hacer mejor. Lo invitamos a seguir apoyándolo."
    else:
        return f"El promedio de {nombre} es menor de 6. Le recomendamos reforzar los temas con actividades adicionales y acercarse a tutorías para mejorar su rendimiento académico."

def replace_recommendation_paragraph(doc, nuevo_texto):
    original_start = "Si el promedio de su hijo(a) es mayor de 9"
    for i, paragraph in enumerate(doc.paragraphs):
        if original_start in paragraph.text:
            doc.paragraphs[i].insert_paragraph_before(nuevo_texto)
            p = doc.paragraphs[i+1]._element
            p.getparent().remove(p)
            break

def crear_grafica(materias, calificaciones, nombre_alumno):
    """Genera una gráfica de barras y la guarda como imagen temporal"""
    plt.figure(figsize=(6, 4))  # Figura más pequeña
    bars = plt.bar(materias, calificaciones, color='skyblue', edgecolor='navy')
    plt.ylim(0, 10)
    plt.ylabel('Calificación', fontsize=10)
    plt.title(f'Rendimiento académico de {nombre_alumno}', fontsize=12)
    plt.axhline(y=6, color='red', linestyle='--', label='Mínimo aprobatorio (6)')
    plt.legend(fontsize=9)
    plt.grid(axis='y', linestyle='--', alpha=0.7)

    # Etiquetas en las barras
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2., height + 0.2,
                 f'{height:.1f}', ha='center', va='bottom', fontsize=9)

    # Rotar etiquetas del eje X si es necesario
    plt.xticks(fontsize=9)
    plt.yticks(fontsize=9)

    # Guardar imagen
    img_path = f"temp_{nombre_alumno.replace(' ', '_')}.png"
    plt.savefig(img_path, dpi=120, bbox_inches='tight')
    plt.close()
    return img_path

def insertar_imagen_despues_de_tabla(doc, imagen_path):
    """Inserta la imagen después de la primera tabla del documento"""
    if doc.tables:
        table = doc.tables[0]
        tbl = table._element
        p = doc.add_paragraph()
        # Insertar la imagen con tamaño reducido (4 pulgadas de ancho)
        p.add_run().add_picture(imagen_path, width=Inches(4))
        tbl.addnext(p._element)
    else:
        doc.add_paragraph().add_picture(imagen_path, width=Inches(4))

# Generar cartas
for idx, row in df.iterrows():
    doc = Document(TEMPLATE_FILE)

    # Ajustar fuente a tamaño 10
    set_document_font(doc, font_size=10)

    # Reemplazar fecha
    replace_text_in_doc(doc, {"23/03/2026": fecha_actual})

    # Rellenar tabla
    fill_table_row(doc, row)

    # Reemplazar recomendación
    promedio = row["Promedio"]
    nombre = row["Alumno"]
    recomendacion = get_recommendation_paragraph(promedio, nombre)
    replace_recommendation_paragraph(doc, recomendacion)

    # Generar gráfica
    materias = ["Español", "Matemáticas", "Ciencias", "Historia", "Biología"]
    calificaciones = [row[m] for m in materias]
    img_path = crear_grafica(materias, calificaciones, nombre)

    # Insertar gráfica después de la tabla
    insertar_imagen_despues_de_tabla(doc, img_path)

    # Guardar documento
    filename = f"Carta_{nombre.replace(' ', '_')}.docx"
    filepath = os.path.join(OUTPUT_DIR, filename)
    doc.save(filepath)
    print(f"Generada: {filepath}")

    # Eliminar imagen temporal
    if os.path.exists(img_path):
        os.remove(img_path)

print(f"\n✅ Proceso completado. Las cartas están en la carpeta '{OUTPUT_DIR}'.")
```
